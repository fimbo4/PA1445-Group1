```python
import matplotlib as plt
from database import vexDB
from vex_file_stats import openvex_analysis

db = vexDB()
pie_data = openvex_analysis(db)

labels = []
sizes = []

for x, y in pie_data.items():
    labels.append(x)
    sizes.append(y)
    
plt.pie(sizes, labels=labels)
plt.axis("equal")
plt.show()
```
