```python
import matplotlib.pyplot as plt
from database import vexDB
from vex_file_stats import repository_categorizer

db = vexDB()
pie_data, num_files = repository_categorizer(db, "OpenVEX")


labels = ["misc"]
sizes = [0]
misc_thresh = 5

for x, y in pie_data.items():
    if y < misc_thresh:
        sizes[0] += y
    else:
        labels.append(x)
        sizes.append(y)

plt.figure(figsize=(10,10))   
plt.pie(sizes)
plt.legend(labels, loc='upper left', bbox_to_anchor=(1, 1))
plt.text(1.0, -1.0, f"Misc threshhold: {misc_thresh}", size="xx-large")
plt.text(-1.0, 1.0, f"No. files:{num_files}", size="xx-large")
plt.show()
```
