```python
import matplotlib.pyplot as plt
import numpy as np
from main import iterate_all_search_terms

BAR_WIDTH = 5
BAR_HEIGHT = 10

vex_filetype_count = iterate_all_search_terms(
    download=False, get_history=False, add_to_db=False, include_optional=False
)

x = []
y = []

for item in vex_filetype_count:
    for key in item.keys():
        x.append(key)
        y.append(item[key])

x = np.array(x)
y = np.array(y)

plt.figure(figsize=(len(vex_filetype_count) * BAR_WIDTH, BAR_HEIGHT))
bar = plt.bar(x, y)
plt.bar_label(bar, fmt=lambda y: f"{y}")
plt.show()

print(vex_filetype_count)

vex_filetype_count = []
```
